```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, MinMaxScaler
```

```python
# Example Dataset

numeric_data = pd.DataFrame({
    'Salary': [30000,45000,60000,80000,120000],
    'Age': [25,32,47,51,62]
})
numeric_data
```

```python
#Standard Scaler
std_scaler = StandardScaler()
std_scaler = std_scaler.fit_transform(numeric_data)
df = pd.DataFrame(std_scaler, columns = numeric_data.columns)
print("\nStandard Scaled Data:\n ", df)
```

```python
minmax_scaler = MinMaxScaler()
minmax_scaler = minmax_scaler.fit_transform(numeric_data)
df1 = pd.DataFrame(minmax_scaler, columns = numeric_data.columns)
print("Min-Max Scaled Data:\n", df1)
```

## Encoding Categorical variables

```python
# converts text label to numeric format for Ml Models
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
```

```python
#Sample Data

cat_data = pd.DataFrame({
    "Department":["HR", "Sales", "IT", "HR", "IT"]
})

cat_data
```

```python
#LabelEncoding
label_encoder = LabelEncoder()
cat_data['Dept_Label'] = label_encoder.fit_transform(cat_data["Department"])
print("\nLabel Encoder:\n", cat_data)
```

```python
one_hot = OneHotEncoder()
cat_data['Dept_One_Hot_Label'] = label_encoder.fit_transform(cat_data["Department"])
print("\nLabel OneHot Encoder:\n", cat_data)
```

```python
cat_data_ohe = pd.get_dummies(cat_data['Department'], prefix = 'Dept').astype(int)
print("\One-Hot Encoded:\n\n", cat_data_ohe)
```

## Normalization

```python
from sklearn.preprocessing import Normalizer

# Example data

X = np.array([[3.0,4.0],[1.0,0.0],[0.0,8.0]])
print("Original Data:\n",X)
normalizer = Normalizer()
normalized_data = normalizer.fit_transform(X)
print("Normalized Data:\n", normalized_data)
```

## Case Study: Retail Customer Data

### Scenario : You have a retail customer dataset with income, age and gender You want to prepare it for ML model input

```python
raw_data = pd.DataFrame({
    'CustomerID': [101,102,103,104,105],
    'Gender': ['Male','Female','Female','Male','Female'],
    'Age': [23,45,31,35,52],
    'Income': [25000,52000,38000,45000,61000]
})
raw_data
```

```python
data = raw_data.drop(columns = "CustomerID")
data
```

```python
label_coder = LabelEncoder()
data['Gender'] = label_coder.fit_transform(data['Gender'])
data
```

```python
from sklearn.preprocessing import  MinMaxScaler
minmaxscaler= MinMaxScaler()
scaled = minmaxscaler.fit_transform(data[['Age','Income']])
data[['Age','Income']] = scaled
print("\nNormalized data\n", data)
                                            
                                                
```
