# Step 5.6: Transforming Skewed Features

Box-Cox transformation helps normalize data and reduce outlier impact, but only works for positive values.

To avoid data leakage, I'll split data, apply Box-Cox transformation parameters derived from the training set to both training and test sets, and use cross-validation for hyperparameter tuning.

```python
# Define the features (X) and the output labels (y)
X = df_encoded.drop('target', axis=1)
y = df_encoded['target']
```

```python
# Splitting data into train and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0, stratify=y)
```

To apply Box-Cox to 'oldpeak', I'll add a small constant (0.001) to ensure all values are positive.

```python
# Adding a small constant to 'oldpeak' to make all values positive
X_train['oldpeak'] = X_train['oldpeak'] + 0.001
X_test['oldpeak'] = X_test['oldpeak'] + 0.001
```

```python
# Checking the distribution of the continuous features
fig, ax = plt.subplots(2, 5, figsize=(15,10))

# Original Distributions
for i, col in enumerate(continuous_features):
    sns.histplot(X_train[col], kde=True, ax=ax[0,i], color='#ff826e').set_title(f'Original {col}')


# Applying Box-Cox Transformation
# Dictionary to store lambda values for each feature
lambdas = {}

for i, col in enumerate(continuous_features):
    # Only apply box-cox for positive values
    if X_train[col].min() > 0:
        X_train[col], lambdas[col] = boxcox(X_train[col])
        # Applying the same lambda to test data
        X_test[col] = boxcox(X_test[col], lmbda=lambdas[col])
        sns.histplot(X_train[col], kde=True, ax=ax[1,i], color='red').set_title(f'Transformed {col}')
    else:
        sns.histplot(X_train[col], kde=True, ax=ax[1,i], color='green').set_title(f'{col} (Not Transformed)')

fig.tight_layout()
plt.show()
```

Inference: Box-Cox transformation improved normality for 'age', 'trestbps', and 'chol', maintained symmetry for 'thalach', and somewhat improved 'oldpeak'.

Transforming features to be more normal-like helps distance-based algorithms by mitigating outlier impact and improving distance calculations.

# Step 6: Decision Tree Model Building

# Step 6.1: DT Base Model Definition

```python
# Define the base DT model
dt_base = DecisionTreeClassifier(random_state=0)
```

I'll tune hyperparameters to achieve high recall for the positive class (heart disease patients), as it's crucial in medical diagnosis.

Note: For medical diagnosis, high recall for the positive class is crucial to avoid missing cases, even if it means some false positives.

Okay, manually tuning all these hyperparameters is going to be a nightmare! I really need a reusable function to automate this, especially focusing on getting the best recall for class 1 (heart disease patients).

```python
def tune_clf_hyperparameters(clf, param_grid, X_train, y_train, scoring='recall', n_splits=3):
    '''Optimizes classifier hyperparameters using GridSearchCV and StratifiedKFold for best recall (class 1).'''

    # Create StratifiedKFold for balanced cross-validation.
    # StratifiedKFold ensures balanced class distribution across folds for imbalanced data.
    cv = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=0)

    # Create the GridSearchCV object
    # n_jobs=-1 means use all available processors, which speeds things up a lot!
    # Maximize recall for class 1 (heart disease) as false negatives are critical.
    clf_grid = GridSearchCV(clf, param_grid, cv=cv, scoring=scoring, n_jobs=-1)

    # Fit the GridSearchCV object to the training data
    # This step can take a while, especially with large grids or complex models!
    clf_grid.fit(X_train, y_train)

    # Get the best hyperparameters
    best_hyperparameters = clf_grid.best_params_

    # Return the best fitted model from GridSearchCV.
    # This is cool because it automatically gives me the best model ready to use!
    return clf_grid.best_estimator_, best_hyperparameters
```

I'll set up the hyperparameters grid and use the `tune_clf_hyperparameters` function to find the optimal settings for my Decision Tree model.

```python
# Hyperparameter grid for DT
param_grid_dt = {
    'criterion': ['gini', 'entropy'],
    'max_depth': [2,3],
    'min_samples_split': [2, 3, 4],
    'min_samples_leaf': [1, 2]
}
```

```python
# Call the function for hyperparameter tuning
best_dt, best_dt_hyperparams = tune_clf_hyperparameters(dt_base, param_grid_dt, X_train, y_train)
```

```python
print('DT Optimal Hyperparameters: \n', best_dt_hyperparams)
```

I'll evaluate the Decision Tree model's performance on both training and test data using classification reports.

```python
# Evaluate the optimized model on the train data
print(classification_report(y_train, best_dt.predict(X_train)))
```

```python
# Evaluate the optimized model on the test data
print(classification_report(y_test, best_dt.predict(X_test)))
```

The model doesn't appear to be overfitting, as training and test metric values are closely aligned.

I'll create a function to consolidate model metrics into a DataFrame for easier comparison later.

```python
def evaluate_model(model, X_test, y_test, model_name):
    """Evaluates trained model performance on test data with various metrics."""
    # Make predictions
    y_pred = model.predict(X_test)

    # Get classification report
    report = classification_report(y_test, y_pred, output_dict=True)

    # Extracting metrics
    metrics = {
        "precision_0": report["0"]["precision"],
        "precision_1": report["1"]["precision"],
        "recall_0": report["0"]["recall"],
        "recall_1": report["1"]["recall"],
        "f1_0": report["0"]["f1-score"],
        "f1_1": report["1"]["f1-score"],
        "macro_avg_precision": report["macro avg"]["precision"],
        "macro_avg_recall": report["macro avg"]["recall"],
        "macro_avg_f1": report["macro avg"]["f1-score"],
        "accuracy": accuracy_score(y_test, y_pred)
    }

    # Convert dictionary to dataframe
    df = pd.DataFrame(metrics, index=[model_name]).round(2)

    return df
```
