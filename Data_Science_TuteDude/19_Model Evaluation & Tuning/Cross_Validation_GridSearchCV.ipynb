# Model Evaluation & Tuning : GridSearchCV

```python
#Import necessary library
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
```

```python
#Load Dataset

iris = load_iris()
X,y = iris.data, iris.target
X[:5]
```

```python
y[:5]
```

```python
#Pipeline : Scaling + Classifier
pipe = Pipeline([
    ("scaler", StandardScaler()),
    ("clf", LogisticRegression(max_iter = 500))
])
```

```python
#Define Hyperparameter Grid

param_grid = {
    "clf__penalty" : ["l1","l2"],
    "clf__C" : [0.01,0.1,1,10],
    "clf__solver": ["liblinear"]
}
```

```python
#GridSearch setup

grid = GridSearchCV(
    estimator = pipe,
    param_grid = param_grid,
    cv =5,
    scoring = "accuracy",
    refit = True,
    n_jobs = -1
)
```

```python
#Run the grid Search
grid.fit(X,y)

print("Best Parameters:", grid.best_params_)
print("Best Cross-validation Accuracy:", grid.best_score_)
print("Test Accuracy with Best Model:", grid.score(X,y))
```

## Python Code Example 1 : Logistic Regression(simple)

```python
#Import necessary library
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
```

```python
#Load Dataset

iris = load_iris()
X,y = iris.data, iris.target
X[:5]
```

```python
y[:5]
```

```python
#Pipeline : Scaling + Classifier
pipe = Pipeline([
    ("scaler", StandardScaler()),
    ("clf", LogisticRegression(max_iter = 500))
])
```

```python
#Define Hyperparameter Grid

param_grid = {
    "clf__penalty" : ["l1","l2"],
    "clf__C" : [0.01,0.1,1,10],
    "clf__solver": ["liblinear"]
}
```

```python
#GridSearch setup

grid = GridSearchCV(
    estimator = pipe,
    param_grid = param_grid,
    cv =5,
    scoring = "accuracy",
    refit = True,
    n_jobs = -1
)
grid.fit(X,y)
```

```python
print("Best Parameters:", grid.best_params_)
print("Best Cross-validation Accuracy:", grid.best_score_)
print("Test Accuracy with Best Model:", grid.score(X,y))
```

```python
##Example 2 - RandomForest (practical, Breast Cancer Dataset)

from sklearn.datasets import load_breast_cancer
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import StratifiedKFold, GridSearchCV
```

```python
#Load Datasets
data = load_breast_cancer()
X,y = data.data, data.target
X[:5]
```

```python
#Define base Model

rf = RandomForestClassifier(random_state=42)

#Define hyperparameter grid to tune
param_grid = {
    "n_estimators" :[50,100,200],
    "max_depth": [None, 5,10],
    "min_samples_split":[2,5,10],
    "min_samples_leaf":[1,2,4]
}
```

```python
#Define CS Strategy
cv = StratifiedKFold(
    n_splits=5, 
    shuffle = True, 
    random_state =42
)
```

```python
#GridSearch setup

grid = GridSearchCV(
    estimator = rf,
    param_grid = param_grid,
    cv =cv,
    scoring = "roc_auc",
    refit = True,
    n_jobs = -1,
    verbose =1
)
grid.fit(X,y)
```

```python
print("Best Parameters:", grid.best_params_)
print("Best Mean ROC-AUC(CV):", grid.best_score_)
print("Final  ROC-AUC  with Best Model:", grid.score(X,y))
```

## Observation

* We used GridSearchCV to systematically tune the hyperparameters of a Random Forest model.

* GridSearchCV tested 81 different parameter combinations, each with 5-fold cross-validation, giving a total of 405 fits.

* The best parameters it found were: max_depth=5, min_samples_leaf=2, min_samples_split=10, n_estimators=100.

* With these parameters, the model achieved an excellent mean ROC-AUC of ~0.991 during cross-validation, and after retraining on the entire dataset, the final ROC-AUC reached ~0.999.

## Summary & Takeaways

* **Cross-Validation (Part 1):** Reliable way to evaluate model performance.
* **GridSearchCV (Part 2):** Uses CV internally to find the best hyperparameters.
* **Example with Logistic Regression and Random Forest showed:**
    * How to define parameter grids.
    * How `GridSearchCV` automates training/testing for all combinations.
    * How to interpret the best parameters and final scores.

### Takeaway
Cross-Validation helps us evaluate models, and GridSearchCV builds on that to *tune* them. Together, they are essential tools for building reliable, high-performing machine learning models.

## Cross-Validation vs. GridSearchCV

To understand the workflow used in this project, here is a direct comparison of how the model was evaluated versus how it was tuned:

| Aspect | Cross-Validation (Part 1) | GridSearchCV (Part 2) |
| :--- | :--- | :--- |
| **Purpose** | Evaluate model performance reliably | Tune model hyperparameters systematically |
| **What it does** | Splits dataset into $k$ folds, trains on $k-1$, tests on 1, repeats | Tries all parameter combinations, uses CV for each, finds best |
| **Output** | Mean $\pm$ std of chosen metric (accuracy, recall, etc.) | Best hyperparameter set + best CV score |
| **Focus** | Generalization ability | Optimization of parameters |
| **When to use** | Anytime you want robust evaluation | When model performance depends heavily on hyperparameters |
| **Example** | Logistic Regression evaluated with 10-fold CV | Random Forest tuned with 81 parameter combinations |
