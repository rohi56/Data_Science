## Predicting Sales Based on Store Footfalls Using Simple linear Regression

### Import required Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

### Load and Preview dataset

```python
data = pd.read_csv("General_Stores.csv")
data.head()
```

```python
data.info()
```

```python
data.shape
```

### Define Independent(Footfalls) and Dependent(DailySales) Variables

```python
X = data[['Footfalls']]
Y = data['DailySales']
```

```python
type(X)
```

```python
type(Y)
```

### Visualize data and check relation between X and Y

```python
plt.xlabel("Footfalls")
plt.ylabel("Daily Sales")
plt.scatter(X,Y,color = 'red', marker ='o')
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
model.fit(X,Y)
```

### Print the model parameters

```python
print(f"Coefficient(slope): {model.coef_}")
print(f"Intercept: {model.intercept_}")
```

### Predict sales of with footfall of 20

```python
model.intercept_+ model.coef_*20
```

```python
model.predict([[20]])
```

```python
data_pred = pd.DataFrame({
    "Footfalls": [25,35,45,100,105]
})
data_pred
```

```python
model.predict(data_pred)
```
