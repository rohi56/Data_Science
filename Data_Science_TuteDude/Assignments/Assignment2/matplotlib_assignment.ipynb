### 1. Create a simple line plot for the following data:
  
###  x = [1, 2, 3, 4, 5]
###  y = [10, 15, 25, 30, 50]

```python
x = [1, 2, 3, 4, 5]
y = [10, 15, 25, 30, 50]
```

#### a. Plot the data. 

```python
plt.plot(x,y)
plt.show()
```

#### b. Customize the plot by adding a title, axis labels, and a grid.

```python
plt.plot(x, y)

plt.title("Simple Line Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.grid()

plt.show()
```

### 2. Create a bar graph to represent the marks scored by students in a subject:  
  
###  students = ['John', 'Jane', 'Alice', 'Bob']
###  marks = [75, 85, 60, 90]


```python
students = ['John', 'Jane', 'Alice', 'Bob']
marks = [75, 85, 60, 90]
```

### a. Plot the bar graph

```python
plt.bar(students, marks)
plt.show()
```

### b. Customize colors and add title

```python
plt.bar(students, marks, color=['red', 'blue', 'green', 'orange'])

plt.title("Marks Scored by Students")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```

### 3. Create a pie chart to represent the percentage distribution of a company’s revenue from different regions:  
  
###  regions = ['North America', 'Europe', 'Asia', 'Others']
###  revenue = [45, 25, 20, 10]

```python
regions = ['North America', 'Europe', 'Asia', 'Others']
revenue = [45, 25, 20, 10]
```

###   a. Create a pie chart with the region names as labels. 

```python
plt.pie(revenue, labels=regions)
plt.show()
```

### b. Highlight the region with the highest revenue.

```python
explode = [0.1, 0, 0, 0]  # Highlight North America (highest value)

plt.pie(revenue, labels=regions, explode=explode, autopct='%1.1f%%')

plt.title("Revenue Distribution by Region")

plt.show()
```

#### 4. Generate a histogram to show the frequency distribution of randomly generated integers between 1 and 100 (sample size = 1000).  

```python
data = np.random.randint(1, 101, 1000)
```

```python
# Create the histogram
plt.hist(data, bins=10)

plt.title("Histogram of Random Integers (1–100)")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()
```
