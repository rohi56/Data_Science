# Used Car Price Analysis Using Matplotlib and Seaborn

## DataSet: Used Car Listing from CarDekho

### Objective: Solve business problems visually - Price trends, feature effects, resale value patterns. 

#### Step 1: Imports & Configuration

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set Theme

sns.set(style = 'whitegrid')
plt.rcParams['figure.figsize'] = (12,8)
```

#### Step 2: Load the Dataset

```python
# Load Dataset

df = pd.read_csv("CAR_DETAILS_FROM_CAR_DEKHO.csv")
df.head()
```

#### Step 3: Fuel Type Distribution

```python
# fule type availability

sns.countplot(x = 'fuel', hue='fuel', data = df, palette ='Set2', legend = True)
plt.title('Fule type distribution')
plt.show()
```

```python
df['fuel'].value_counts()
```

#### Observations:

- **Petrol and Diesel dominate the market**: Both fuel types have overwhelmingly higher counts compared to others, indicating that most used cars listed are either petrol or diesel vehicles. Diesel appears slightly higher than petrol, suggesting a marginal preference or higher resale presence.
- **Alternative fuels are rare**: CNG and LPG vehicles make up only a very small fraction of the listings. This shows limited adoption or availability of these fuel types in the used car market.
- **Electric vehicles are nearly absent**: The count for electric cars is extremely low (almost negligible), indicating that electric vehicles have not yet significantly penetrated the used car segment in this dataset.

### Overall Inference:
The used car market is still heavily reliant on traditional internal combustion engine vehicles (petrol and diesel), with alternative and electric fuel types playing only a minimal role.

#### Step 4 : Selling price distribution

```python


# price range analysis

sns.histplot(df['selling_price'], kde= True, bins= 30, color = 'skyblue')
plt.title("Selling price distribution")
plt.xlabel("Price (INR)")
plt.show()
             
             
```

#### Observation:

- **Highly right-skewed distribution**: Most cars are concentrated in the lower price range, with the frequency rapidly decreasing as price increases. This indicates that affordable cars dominate the used car market.
- **Majority of listings are low to mid-priced**: A large number of cars fall roughly below ₹5–10 lakhs, suggesting that buyers and sellers are more active in this price segment.
- **Few high-value outliers**: There are very few cars with extremely high prices (extending toward ₹50–90 lakhs), creating a long tail on the right side of the distribution. These are likely luxury or premium vehicles.
- **Peak density at lower prices**: The highest density is observed at the lower end of the price spectrum, reinforcing that most transactions occur in the budget-friendly segment.

#### Overall Inference:

The used car market is dominated by lower-priced vehicles, with only a small proportion of premium or luxury cars. This suggests strong demand and supply in the affordable segment, while high-end vehicles are relatively rare.

#### Step 5: Price by fuel Type (Boxplot)

```python
# Fuel vs Price
sns.boxplot(x = 'fuel', hue = 'fuel', y = 'selling_price', data = df, palette = 'Set1')
plt.title("Price Distribution by Fuel Type")
plt.show()
```

#### Observation:

- **Diesel cars have higher median prices**: Diesel vehicles generally show a higher median selling price compared to petrol, indicating they tend to retain more value or belong to slightly higher segments.
- **Petrol cars are moderately priced**: Petrol vehicles have a lower median than diesel but still show a wide range, making them common across budget to mid-range categories.
- **CNG and LPG are lower-priced segments**: Both CNG and LPG cars have relatively lower median prices and tighter distributions, suggesting they are mostly budget-friendly vehicles with less variability.
- **Electric cars show limited data**: Electric vehicles appear with very few data points, making it hard to generalize, but they seem to be priced moderately higher than CNG/LPG.
- **Diesel shows highest variability and outliers**: Diesel cars have a wide spread and many high-price outliers (luxury SUVs, premium vehicles), indicating strong presence across both mid-range and high-end markets.
- **Petrol also has high-value outliers**: Though fewer than diesel, petrol cars include some very expensive listings, contributing to a broader price spread.

#### Overall Inference:

Fuel type significantly influences pricing in the used car market. Diesel cars tend to dominate the higher price segments with greater variability, while petrol cars serve a wide range of buyers. CNG and LPG vehicles are mostly confined to the lower-price segment, and electric vehicles are still too few to establish a clear pricing trend.

#### Price By Transmission

```python
# Transmission Impact

sns.boxplot(x = 'transmission', hue = 'transmission' , y = 'selling_price' , data = df, palette = 'coolwarm')
plt.title('Selling Price by Transmission Type')
plt.show()
```

#### Observation:

- **Automatic cars have higher median prices**: The median selling price for automatic vehicles is significantly higher than for manual ones, indicating that automatics are generally positioned in more premium segments.
- **Manual cars dominate the lower price range**: Manual vehicles show a lower median and a tighter price range, suggesting they are more common in budget and mid-range categories.
- **Greater price variability in automatic cars**: Automatic vehicles exhibit a much wider spread (larger interquartile range), indicating diversity from mid-range to luxury cars.
- **High-value outliers are mostly automatic**: The most expensive cars (extreme outliers) are predominantly automatic, reinforcing that luxury and high-end vehicles are usually equipped with automatic transmission.
- **Manual cars still have some outliers**: While fewer than automatics, manual cars do show some higher-priced outliers, but these are relatively limited.

#### Overall Inference:

Transmission type has a strong impact on pricing. Automatic cars are generally more expensive and span a broader price spectrum, often representing premium and luxury vehicles. In contrast, manual cars are more concentrated in the affordable segment, making them the dominant choice for budget-conscious buyers.

#### Step 7 : Price by Ownership History

```python
# Owner Impact

sns.violinplot(x = 'owner', hue = 'owner', y = 'selling_price', data = df, palette = 'Set3')
plt.title('Selling Price vs Owner History')
plt.show()
```

#### Observation:

- First-owner cars have higher prices on average.
- Price decreases as the number of owners increases.
- Fourth & above owner cars are the cheapest.
- Test drive cars are priced higher, likely due to being nearly new.

Overall: Fewer owners = higher value; more owners = lower resale price.

#### Step 8: Correlation Heatmap

```python
correlation = df[['year', 'km_driven','selling_price']].corr()

sns.heatmap(correlation, annot = True, cmap = 'YlGnBu')
plt.title('Feature Correlation Heatmap')
plt.show()
```

#### Observations:

- **Selling Price and Year (Positive Correlation: 0.41)**:
There is a moderate positive correlation between the manufacturing year and the selling price. This indicates that newer cars (higher year value) tend to command higher selling prices. As the car's age decreases, its market value generally increases.

- **Year and Kilometers Driven (Negative Correlation: -0.42)**:
There is a moderate negative correlation between the year and the kilometers driven. This is logical: older cars (lower year value) typically have higher mileage, while newer cars have been on the road for less time and thus have fewer kilometers on the odometer.

- **Selling Price and Kilometers Driven (Negative Correlation: -0.19)**:
There is a weak negative correlation between the distance driven and the price. While you might expect a stronger link, this suggests that while higher mileage does lower the price, other factors (like the brand, fuel type, or the specific year) might be playing a more dominant role in determining the final selling price than mileage alone.
