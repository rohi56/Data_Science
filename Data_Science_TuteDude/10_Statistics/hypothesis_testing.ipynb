# Hypothesis Testing for Loyalty Program Launch for supermarket

## Description

**Bussiness Question:** Should he supermarket launch a loyalty program?

**Decision Criteria:** Launch only if the average spending > $120/week per shopper.

**Pilot Program Results:**

- Sample Size : 80 shoppers
- Sample mean : $130/Week
  
- Sample standard Deviation: $ 40

**Statistical Question:** Is the observerd $130 average significantly greater than the USD threshold?

```python
# Import required library
import scipy.stats as stats
```

```python
# Define the business and statistical inputs:
## Sample size (number of shopper in the pilot program)

n = 80

# Sample mean weekly spending from the pilot program
sample_mean = 130 #dollar/week

#sample standard deviation of weekly spending
sample_std = 40 #dollar/week

# hypothesized population mean (status quo threshold)
mu_0 = 120
```

## Step 1: Setup the hypothesis

- **Null Hypothesis(H0):** The true average spending is < = $120.
  
- **Alternative Hypothesis(H1):** The true average spending is > $120

## Step 2: Calculate the t-statistics

**Formula:** t = (sample_mean - hypothesized_mean) / (sample_std /sqrt(n))


```python
t_statistic = ( sample_mean - mu_0)/ (sample_std / (n ** 0.5))
print(t_statistic)
```

Since the business decision depends on whether the average spending is greater tha $ 120,

we perform a right-tailed one-sample t-test

to test if the sample mean ($130) is significantly higher than the threshold.

Hint to decide:

- If the question uses "greater than", it's a right-tailed test.
- If it's "less than", it's a left tailed test.
- If it's "different from", it's a two tailed test.

## Step 3: Calculate the p-value from t value

Since we're testing if the mean is Greater than $120, it's a one tailed test

```python
p_value = 1-stats.t.cdf(t_statistic, df= n-1)
print(p_value)
```

## Step 4 : Choose a significance level (commonly 0.05 = 5%)

```python
alpha = 0.05
```

## Step 5: Make the Decision

```python
if p_value < alpha:
    decision = "Reject the null hypothesis : Launch the loyalty program."
else: 
    decision = "Fail to reject the null hypothesis : Do not launch the program."

print(decision)
```

## Display the Value 

```python
print(f"T-Statistic: {t_statistic:.2f}")
print(f"P-Value:{p_value:.4f}")
print("Decision:", decision)
```
