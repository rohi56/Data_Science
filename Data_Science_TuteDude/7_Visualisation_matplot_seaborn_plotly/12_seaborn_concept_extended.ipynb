## Relational Plots

### Defination:

**Relational Plots** show relationships between **two more numerical Variables.**

### Usage:

- Used to visualize relationships in datasets(e.g correlation between age and medical expenses).
- Helps understand **trends, patterns and clusters** in numerical data.

### Characteristics:

- Typically uses **scatter plots(`sns.scatterplot`) and line plots(`sns.lineplot`)**.
- Supports **hue,size and style** to add more insights.

### Applications

- **Telecomunication:** Relationship between **call duration and customer satisfaction**.
- **Finance:** Relationship between **income and expenditure.**
- **Retail:** Impact of **discount percentage on sales**.

# Telecommunication Case Study: Customer Call Analytics

## Problem Defination

The objective is to analyze **customer calling behaviour** to understand trends in call duration, 

network complaints and customer enagagement patterns.

## Dataset

- **Call_Duration:** Duration of each call(in minutes)
- **Network_Provider:** Telecom provider (e.g. Provider A, Provider B)
- **Satisfaction_score:** Customer's satisfaction score(1-10)

## Relational Plot: Call Duration vs. Satisfaction Score

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df =pd.read_csv('telecom_customer_data.csv')
df.head()
```

```python
df.shape
```

```python
#Create the scatter plot
plt.figure(figsize=(8,5))
sns.scatterplot(data = df, x = "Call_Duration", y = 'Satisfaction_Score', hue = "Network_Provider")

#Move the legend outside the plot

plt.legend(title= "Network Provider", bbox_to_anchor=(1,1), loc ="upper left")

### Add labels and title

plt.xlabel("Call Duration (minutes)")
plt.ylabel("Satisfaction Score")
plt.title("Call Duration vs. Satisfaction Score")

# Show plot
plt.show()
```

## What is your observation/Inference?

Longer call durations may correlate with lower satisfaction

shorter call duration found more satisfactory, between upto 25 minutes particularly.

## Categorical Plots

### Defination:

Categorical plots display relationship involving **categorical Variables**(e.g., gender, department, product category).

### Usage:

- Used to analyze **grouped data**(e.g., average number of call per customer segment).
- Helps in comparing different categories.

### Characteristics:

- Include **bar plot(`sns.barplot`), box plot(`sns.boxplot`), violin plots(`sns.violinplot`)** etc.
- Allows **comparison across multiple groups.

### Applications: 

- **Telecommunication:** Distribution of **customer complaints by network provider**.
- **Healthcare:** Distribution of **patient recovery time by disease**
- **Eductation:** Comparison of **student's performance by subject**.

### Categorical Plot: Count of Customer by Network Provider

```python
plt.figure(figsize = (10,5))
sns.countplot(data =df, x = 'Network_Provider', hue = 'Network_Provider', palette = 'Set2')
plt.title("Customer Distribution by Network Provider")
plt.xlabel("Network Provider")
plt.ylabel("Count of Customers")

plt.show()
```

## What is your observation/Inference?

Provider A has highest customers which is approximately 180, which is 10% higher tha provider B, lowest is provider C.


## Distribution Plots

### Defination:

Distribution plots visualize the **Spread and shape of data**.

### Usage:** 

- Helps to understand **data distribution, skweness, and outliners.**
- Useful for **identifying trends and anomalies.**

### Characteristics:

- Include **histogram(`sns.histplot`), kde plot(`sns.kdeplot`) and rug plots(`sns.rugplot`).**
- Display **density estimates and probability distributions.**

### Applications:**

- **Telecomunication:** Distribution of **call durations per customers.**
- **Finance:** Distribution of **transaction amount in a bank.**
- **E-commerce:** Analyzing **purchase frequency of products.**

### Distribution Plots: Distribution of call Durations

```python
plt.figure(figsize = (10,5))
sns.histplot(df['Call_Duration'], bins = 20, kde = True, color = 'blue')
plt.title("Distribution of Call Duration")
plt.xlabel("Call Duration (minutes)")
plt.ylabel("Frequency")
plt.show()
```

## What is your observation/Inference?

Shows the most common call durations, which uniforms and Frequency are highest 35 at call duration between 30-40 otherwise it found around average and
may be very few outliners.


##### Note: Outlier is far away from mean

## Matrix Plots

### Defination:

Matrix plots visualize **relationship between multiple variables** using **heatmaps and clustering methods**.

### Usage:

- Helps find correlation and patterns in datasets.
- Used in **feature selection and anomaly detection.**

### Characterstics:

- Include **heatmaps(`sns.heatmap`) and clustermaps(`sns.clustermap`)**
- Display **intensity and similarity between variables.**

### Applications:

- **Telecommunication:** Heatmap of **network congestion across different locations.**
- **Finance:** Correlation between **various stock indices.**
- **Weather Analytics:** Heatmap of **temperature variation over time.**

### Matrix plot: Correlation Matrix of Features"

```python
plt.figure(figsize = (8,6))

# Compute corelation for numeric columns only

numeric_df = df.select_dtypes(include = ['number']) # Select only numeric columns
sns.heatmap(numeric_df.corr(), annot = True, cmap = 'coolwarm', fmt = ".2f", linewidths = 0.5)

plt.title("Correlation Matrix of Features")
plt.show()
```

## What is your observation/Inference?

- The correlation between call duration and satisfaction score is -0.75, including a strong negative correlation.
- This means that as call Duration Increases, Satisfaction Score trends to decreases.
    




```python
plt.figure(figsize = (8,5))
sns.boxplot(data =df, x = 'Network_Provider', y = 'Satisfaction_Score')
plt.xlabel("Network Provider")
plt.ylabel("Satisfaction Score")
plt.title("Satisfaction score across different network providers")
plt.show()
```

## What is your observation/Inference?

- Higher medain satisfaction for a provider may iondicate better service quality.
- Wide IQR(Interquartile Range) means user experiences vary greatly.
- Outliners suggest positive or negative experiences.




### Violin Plot: Call Duration by Network Provider

Swarmplots are useful for displaying individual data points, especially when dealing with categorical variables.

```python
plt.figure(figsize = (8,5))
sns.violinplot(data = df, x = 'Network_Provider', y = 'Call_Duration')
plt.xlabel("Network Provider")
plt.ylabel("Call Duration (minuets)")
plt.title("Call Duration Distribution Across Network Providers")
plt.show()
```

## What is your observation/Inference?

- Provides with wide distributions have highly varing call durations.
- Narrorwer shapes indicates more consistent call durations.
- Long tails suggest some customers have significantly longer or shorter calls.
