#### 1. Create a 1D NumPy array with integers from 1 to 20. Perform the following operations: 

```python
arr1 = np.arange(1,21)
print(arr1)
```

####   a. Calculate the sum, mean, median, and standard deviation of the elements in the array. 

```python
sum_arr1 = np.sum(arr1)
print("Sum of arr1:", sum_arr1)
```

```python
mean_arr1 = np.mean(arr1)
print("Mean of arr1:", mean_arr1)
```

```python
median_arr1 = np.median(arr1)
print("Median of arr1:", median_arr1)
```

```python
std_arr1 = np.std(arr1)
print("Standard Deviation of arr1:", std_arr1)
```

####  b. Find the indices of elements greater than 10 in the array.

```python
greater_than_10 = arr1[arr1 > 10]
print(f" Indices of elements greater than 10 in arr1:\n{greater_than_10}")
```

### 2. Create a 2D NumPy array of shape 4 X 4 with numbers ranging from 1 to 16.

```python
arr2 = np.arange(1, 17).reshape(4, 4)
```

#### a. Print the array.

```python
print(f"2D NumPy array of shape 4 X 4 with numbers ranging from 1 to 16:\n{arr2}")
```

####  b. Find the transpose of the array.

```python
arr2_t = np.transpose(arr2)
print(f"The transpose of the arr2 is :\n{arr2_t}")
```

#### c. Calculate the row-wise and column-wise sums of the array.

```python
sum_of_row = np.sum(arr2, axis = 1)
print(f"Sum of row of arr2: {sum_of_row}")
```

```python
sum_of_column = np.sum(arr2, axis = 0)
print(f"Sum of column of arr2: {sum_of_column}")
```

### 3. Create two 3 X 3 arrays filled with random integers between 1 and 20.

```python
rand_arr1 = np.random.randint(1,20, (3,3))
print(rand_arr1)
```

```python
rand_arr2 = np.random.randint(1,20, (3,3))
print(rand_arr2)
```

#### a. Perform element-wise addition, subtraction, and multiplication.

```python
print(f"element-wise addition:\n {rand_arr1 + rand_arr2}")
```

```python
print(f"element-wise subtraction:\n {rand_arr1 - rand_arr2}")
```

```python
print(f"element-wise multiplication:\n {rand_arr1 * rand_arr2}")
```

#### b. Compute the dot product of the two arrays.

```python
dot_rand = np.dot(rand_arr1, rand_arr2)
print(f"dot product of the two arrays:\n {dot_rand}")
```
