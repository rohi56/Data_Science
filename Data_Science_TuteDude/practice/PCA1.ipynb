# Dimentionality Reduction Using PCA **Principal Component Analysis** on Iris Dataset

```python
#Import Required Library

import numpy as np
import matplotlib.pyplot as plt

from sklearn.decomposition import PCA
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
```

```python
#Load Dataset

iris = load_iris()
X = iris.data
y = iris.target

target_names = iris.target_names
X[:5]
```

```python
target_names
```

```python
y
```

```python
X_scaled = StandardScaler().fit_transform(X)
```

```python
pca = PCA(n_components =4)
X_pca = pca.fit_transform(X_scaled)

X_pca[:5]
```

```python
print("Explained varaince ratio:", pca.explained_variance_ratio_)
```

```python
print("Cumulative variance explained :", np.cumsum( pca.explained_variance_ratio_))
```

# Observation:

Example :First two components may capture ~95% variance -> can reduce from 4d -> 2D

```python
#Scree Plot : Line plot of cumulative explained variance vs number of components

plt.plot(np.cumsum( pca.explained_variance_ratio_), marker = "o")
plt.xlabel("Number of Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("Scree Plot for PCA on Iris Dataset")
plt.grid(True)
plt.show()
```

```python
# Scatter plot of 2D PCA-transformed data

plt.figure(figsize =(8,6))
colors = ["red","green","blue"]
for i, target in enumerate(np.unique(y)):
    plt.scatter(X_pca[y==target, 0], X_pca[y==target,1],
                color = colors[i], label=target_names[i], alpha =0.7)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("2D Visualization on Iris Dataset using PCA")
plt.grid(True)
plt.legend()
plt.show()
```
