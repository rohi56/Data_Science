```python
import numpy as np
import matplotlib.pyplot as plt
```

### Figure Class

#### Defination:

The **figure class** is the overcall container for all the elements in a Matplotlib plot.

It acts as a canvas where multiple plos (Axes) can be placed.

#### Application:

- Used when creating multiple subplots in a single figure.
- Helpful for complex visualizations with different charts in one display.


```python
# Example

fig, ax = plt.subplots()
ax.set_title("Basic Figure and axes")
ax.set_xlabel("Days")
ax.set_ylabel("total patients")
plt.show()
```

### Axes Class

#### Defination:

The **Axes Class** represents individual plots inside a Figure.

Each figure can have multiple Axes, each containing a seperate graph.

#### Application:

- Used when plotting multiple realated visualization togrther (e.g. sales trends vs. product categories in retail analytics).
- Helps control labels, ticks, legends, and other customization for each plot.

### Line Plot

#### Defination:

The **Line Plot** displays data points connected by lines. It is commonly used for showing trends over time.

#### Application:

- Used for time series analysis(e.g monthly sales, website traffic, temperature changes).
- In healthcare, used to show patient recovery trends over weeks.

#### Line Plot: Patient Recovery Trends

##### Problem Statement:

A hospital tracks the recovery rate of patients over several months.

We visualize the trends using a line plot.

```python
months = ['Jan','Feb','Mar','Apr','May','Jun']
recovery_rate = [65,70,60,78,80,85]

plt.plot(months, recovery_rate, marker ='o', linestyle= '-', color = 'b', label = 'Recovery Rate')

plt.xlabel("months")
plt.ylabel("Recovery Percentage")
plt.title("Patient Recovery Trends")
plt.legend()
plt.show()
        
```

#### What is the observation about this line plot?

#### Inference

Trend of recovery percentage is upward from January to June, except for the month of March is downward.

### Bar Plot

#### Defination:

The **Bar Chart** represent categorical data using rectangular bars, where the height represent numerical value.

#### Application:

- Used to compare diddefent categories(e.g sales per product, customer segmentation).
- In healthcare, it can show number of patient in different hospital departments.

#### Bar Plot: Medicine Demand Analysis

##### Problem Statement:

A pharmacy wants to analyze the demand for different medicines.

We use a bar plot to visualize the demand.

```python
medicines = ['Paracetamol', 'Antibiotics','Painkillers','Insulin','Antacids']
demand = [500,700,450,300,650]

plt.bar(medicines, demand, color = ['red','blue','green','purple','orange'])
plt.xlabel("Medicine Type")
plt.ylabel("Units Sold")
plt.title(" Medicine Demand Analysis")

plt.show()
```

#### What is the observation about this Bar plot?

#### Inference

Highest selling is of Antibotics around 700.
Lowest selling is of Insulin around 300, rest are between this range of 300-700


### Pie Chart

#### Defination:

The **Pie Chart** shows proportions of different categories within a whole.

#### Application:

- Used for percentage distributions (e.g market share of different brands).
- In healthcare, it can visualize percentage distribution of disease cases(e.g., diabates, hypertension, heart disease, etc.).

#### Pie Chart: Disease Distribution

##### Problem Statement:

A Healthcare centre analyzed the distribution of disease in a city.

We visualize the distribution using a pie chart.

```python
disease= ['Diabetes', 'Hypertension', 'Heart Disease', 'Asthma','Others']
percentage= [30,25,15,10,20]
colors = ['blue', 'red','green', 'purple','orange']

plt.pie(percentage, labels = disease, autopct ='%1.1f%%', colors = colors)
plt.title("Disease Distribution in City")
plt.show()
```

#### What is the observation about this Pie Chart?

#### Inference

As per the pie chart, Diabates has higgest number of Patients, that is `30%` .
Least is Astama around `10%`.

### Scatter Plot

#### Defination:

The **Scatter Plot** shows individual data points on a 2D plane. It is used to identify relationships between two variables.

#### Application:

- Used for correlation analysis (e.g advertising spends vs. revenue).
- In healthcare, it can show corellation between BMI and Blood Sugar levels.

#### Scatter Plot: Doctor-Patient Interaction Time

##### Problem Statement:

A hospital examines how time spent by doctors affects diagnosis accuracy.
We visualize this relationship with a scatter plot.

```python
time_spent =[10,15,20,25,30,35,40]
diagnosis_accuracy =[60,65,70,75,80,85,90]

plt.scatter(time_spent, diagnosis_accuracy, color = 'red')
plt.xlabel("Time spent by doctor(minutes)")
plt.ylabel("Diagnosis Accuracy (%)")

plt.title("Doctor-Patient Interaction Time Vs. Diagnosis Accuracy")
plt.show()
```

#### What is the observation about this Scatter Plot?

#### Inference

Here trend is upward, that is positive for Diagnosis Accuracy vs Time Spent by Doctor.

### Histogram

#### Defination:

The **Histogram** displays the frequency distribution of a dataset by grouping values into bins.

#### Application:

- Used for distribution analysis analysis (e.g age distribution of customers).
- In healthcare, used to analyze blood pressure distribution among patients.



#### Histogram: Age Distribution of Patients

##### Problem Statement:

We analyze the age distribution of patient vising a hospital using a histogram.



```python
ages = np.random.randint(10,80,100)
plt.hist(ages, bins = 10, color= 'blue', edgecolor = 'black')

plt.xlabel("Age Groups")
plt.ylabel("Number of Patient")
plt.title("Age Distribution of Patients")
plt.show()
```

#### What is the observation about this Histogram?

#### Inference

Here the patient with the most ages are vistied more in hospital.

### Word Cloud

#### Defination:

The **Word Cloud** a visual representation of text data where the size of each word is proportional to its frequency in the dataset.

#### Application:

- Used for text analysis (e.g most common customer complaints).
- In healthcare, used to analyze patients feedback to identify common health concerns.




#### Word Cloud: Common Symptoms Analysis

##### Problem Statement:

A hospital collects common symptoms from patient records.
We visualize frequently occuring symptoms using word cloud

```python
#!pip install wordcloud
```

```python
from wordcloud import WordCloud
```

```python
symptom_text = " fever cough cold headache nausea fatigue pain weakness dizziness allergy infection sore throat congestion"
wordcloud = WordCloud(width = 800, height = 400, background_color = 'white').generate(symptom_text)

plt.figure(figsize = (10,5))
plt.imshow(wordcloud, interpolation = 'bilinear')
plt.axis("off")
plt.title("Common Sympton Analysis")
plt.show()
```

#### What is the observation about this WordCloud?

#### Inference

- It shows highest frequency word is fever, cough then clod.
- This words are frequently used in communication.
- Word like headache, nausea, fatigue, pain are used second highest used.
