# Time Series Forecasting : Naive Forecasting (Simplest Forecasting Method)

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#Dataset

data = {
    "Month":[ "Jan", "Feb", "Mar", "Apr"],
    "Sales" : [200,220,240, None] # April is unknown
}

df = pd.DataFrame(data)
df
```

```python
# Plot existing data

plt.plot(df["Month"], df["Sales"], marker = "o", label = "Actual Sales")
plt.title("Naive Forecast Example")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)

#Naive forecast

naive_forecast = df["Sales"].iloc[2] #last known value (March)
plt.plot("Apr", naive_forecast, marker = "x", markersize =10, color = "red", label="Forecast (Naive)")

plt.legend()
plt.show()

print(f"Naive Forecast for April: {naive_forecast}")
```
