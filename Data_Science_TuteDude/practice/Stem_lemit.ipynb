## Stemming

```python
from nltk.stem import PorterStemmer
```

```python
stemmer = PorterStemmer()
print(stemmer.stem("running"))
print(stemmer.stem("flying"))
print(stemmer.stem("flies"))
print(stemmer.stem("responsibility"))
print(stemmer.stem("runner"))
```

## Lemitization

```python
import nltk
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download("wordnet")
nltk.download("omw-1.4")
```

```python
lemmatizer = WordNetLemmatizer()
print(lemmatizer.lemmatize("running", pos ="v"))
print(lemmatizer.lemmatize("better", pos ="a"))
print(lemmatizer.lemmatize("flies", pos ="v"))
```

## Full Example : Stopword Removal + Lemmatization

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

#Get English stop words
stop_words = stopwords.words("english")

text = "Cats are running around the house, chasing mice happily."

tokens = word_tokenize(text)

filtered = [word for word in tokens if word.lower() not in stop_words]

lemmatized = [lemmatizer.lemmatize(word.lower(), pos = 'v') for word in filtered]
print("Preocessed :", lemmatized)
```
