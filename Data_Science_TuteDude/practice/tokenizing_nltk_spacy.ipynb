```python
import nltk
#nltk.download("punkt_tab")
from nltk.tokenize import word_tokenize, sent_tokenize
```

```python
# Input text

text = "Hello!, I'm Learning NLP. Let's text tokenization -sentence-wise and word wise. Amazing, isin't it?"


sentences = sent_tokenize(text)
```

```python
sentences
```

```python
print("========NLTK Sentences Tokenization=========")
for i, sent in enumerate(sentences):
    print(f"Sentence{i+1}: {sent}")


print("\n=== NLTK Word Tokenization ===")

for i, sent in enumerate(sentences):
    print(f"Sentence{i+1} tokens:")
    word = word_tokenize(sent)
    print(word)
    
```

```python
## spacy tokenization Example
```

```python
import spacy

nlp = spacy.load("en_core_web_sm") #Load the small English Model

doc = nlp(text)

print("\n === spaCy Tokenization === ")
print("Tokens:")
print()
print()

for token in doc:
    print(token.text)

print()
print()
print("\n === spaCy Sentence Segmentation ===")

for i, sent in enumerate(doc.sents):
    print(f"Sentence{i+1}:{sent.text}")
```
